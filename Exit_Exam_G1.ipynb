```python
import pandas as pd
from sklearn.model_selection import train_test_split, GridSearchCV, StratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    f1_score,
)


df.columns = df.columns.str.strip()
df = df.replace(r"^\s*$", pd.NA, regex=True)

target = "Best_Shopping_Mode"
df = df.dropna(subset=[target])

X = df.drop(
    columns=[target, "Response_ID", "Timestamp"],
    errors="ignore",
)
y = df[target].astype(str)

# Numeric columns
numeric_columns = [
    "Age",
    "Review_Influence_Score",
    "Social_Ads_Influence_Score",
    "Price_Comparison_Frequency",
    "Discount_Importance_Score",
    "Likely_To_Buy_After_Ad_Score",
    "Impulse_Purchase_Frequency",
    "Online_Shopping_Satisfaction_Score",
    "Trust_In_Online_Reviews_Score",
    "Regret_After_Purchase_Frequency",
]

numeric_columns = [col for col in numeric_columns if col in X.columns]

for col in numeric_columns:
    X[col] = pd.to_numeric(X[col], errors="coerce")

categorical_columns = [
    col for col in X.columns
    if col not in numeric_columns
]

# Remove completely empty columns
X = X.dropna(axis=1, how="all")

# Stratified split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y,
)

print("Training shape:", X_train.shape)
print("Testing shape:", X_test.shape)
```

```python
# Preprocessing
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False,
    )),
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_columns),
    ("categorical", categorical_pipeline, categorical_columns),
])

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42,
)

models = {
    "Logistic Regression": {
        "classifier": LogisticRegression(
            max_iter=3000,
            class_weight="balanced",
            random_state=42,
        ),
        "parameters": {
            "classifier__C": [0.01, 0.1, 1, 10],
            "classifier__solver": ["lbfgs"],
        },
    },
    "Decision Tree": {
        "classifier": DecisionTreeClassifier(
            class_weight="balanced",
            random_state=42,
        ),
        "parameters": {
            "classifier__max_depth": [3, 5, 10, None],
            "classifier__min_samples_leaf": [5, 10, 15],
        },
    },
    "Random Forest": {
        "classifier": RandomForestClassifier(
            class_weight="balanced",
            random_state=42,
            n_jobs=-1,
        ),
        "parameters": {
            "classifier__n_estimators": [300],
            "classifier__max_depth": [5, 10, None],
            "classifier__min_samples_leaf": [2, 5, 10],
        },
    },
    "Gradient Boosting": {
        "classifier": GradientBoostingClassifier(random_state=42),
        "parameters": {
            "classifier__n_estimators": [100, 200],
            "classifier__learning_rate": [0.03, 0.05, 0.1],
            "classifier__max_depth": [3, 5],
        },
    },
}

comparison_results = []
trained_models = {}

for model_name, model_config in models.items():
    pipeline = Pipeline([
        ("preprocessing", preprocessor),
        ("classifier", model_config["classifier"]),
    ])

    search = GridSearchCV(
        estimator=pipeline,
        param_grid=model_config["parameters"],
        cv=cv,
        scoring="f1_macro",
        n_jobs=-1,
        refit=True,
    )

    search.fit(X_train, y_train)

    predictions = search.predict(X_test)

    comparison_results.append({
        "Model": model_name,
        "Best parameters": search.best_params_,
        "CV macro-F1": search.best_score_,
        "Test accuracy": accuracy_score(y_test, predictions),
        "Test balanced accuracy": balanced_accuracy_score(
            y_test,
            predictions,
        ),
        "Test macro-F1": f1_score(
            y_test,
            predictions,
            average="macro",
        ),
        "Test weighted-F1": f1_score(
            y_test,
            predictions,
            average="weighted",
        ),
    })

    trained_models[model_name] = search

comparison_table = (
    pd.DataFrame(comparison_results)
    .sort_values("CV macro-F1", ascending=False)
)

print(comparison_table.to_string(index=False))

# Select by cross-validation macro-F1
best_model_name = comparison_table.iloc[0]["Model"]
best_model = trained_models[best_model_name]
best_predictions = best_model.predict(X_test)

print(f"\nBest model: {best_model_name}")
print(f"Accuracy: {accuracy_score(y_test, best_predictions):.4f}")
print(f"Balanced accuracy: "
      f"{balanced_accuracy_score(y_test, best_predictions):.4f}")
print(f"Macro-F1: "
      f"{f1_score(y_test, best_predictions, average='macro'):.4f}")
print(f"80% accuracy achieved: "
      f"{accuracy_score(y_test, best_predictions) >= 0.80}")

print("\nClassification report:")
print(classification_report(
    y_test,
    best_predictions,
    zero_division=0,
))

print("Confusion matrix:")
labels = sorted(y.unique())

matrix = confusion_matrix(
    y_test,
    best_predictions,
    labels=labels,
)

print(pd.DataFrame(
    matrix,
    index=labels,
    columns=labels,
))

ConfusionMatrixDisplay(
    confusion_matrix=matrix,
    display_labels=labels,
).plot(cmap="Blues", values_format="d")
```

```python
from sklearn.ensemble import StackingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.pipeline import Pipeline
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.metrics import accuracy_score, classification_report

stacking_classifier = StackingClassifier(
    estimators=[
        ("logistic", LogisticRegression(
            C=0.1,
            class_weight="balanced",
            max_iter=1000,
            random_state=42
        )),
        ("random_forest", RandomForestClassifier(
            n_estimators=300,
            max_depth=5,
            min_samples_leaf=5,
            class_weight="balanced",
            random_state=42,
            n_jobs=-1
        )),
        ("gradient_boosting", GradientBoostingClassifier(
            n_estimators=100,
            learning_rate=0.05,
            max_depth=3,
            random_state=42
        ))
    ],
    final_estimator=LogisticRegression(
        class_weight="balanced",
        max_iter=1000,
        random_state=42
    ),
    stack_method="predict_proba",
    cv=5,
    n_jobs=-1
)

stacking_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", stacking_classifier)
])

stacking_model.fit(X_train, y_train)
predictions = stacking_model.predict(X_test)

print("Test Accuracy:", round(
    accuracy_score(y_test, predictions), 4
))
print(classification_report(
    y_test,
    predictions,
    zero_division=0
))

cv = StratifiedKFold(
    n_splits=10,
    shuffle=True,
    random_state=42
)

cv_scores = cross_val_score(
    stacking_model,
    X,
    y,
    cv=cv,
    scoring="accuracy",
    n_jobs=-1
)

print("Cross-validation Accuracy:",
      round(cv_scores.mean(), 4))
```
